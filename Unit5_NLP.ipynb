```python

!pip install nltk scikit-learn -q

import nltk, random, re
import numpy as np
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('stopwords', quiet=True)
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


corpus = """
Natural Language Processing (NLP) is a subfield of artificial intelligence.
NLP helps computers understand, interpret, and generate human language.
Machine learning is used to train NLP models on large text datasets.
Text classification assigns categories to documents using algorithms.
Named entity recognition identifies names, places, and organizations in text.
Sentiment analysis determines whether text is positive, negative, or neutral.
Word embeddings represent words as dense vectors in a continuous space.
Transformers are a deep learning architecture used widely in modern NLP.
BERT is a transformer model pre-trained on large text corpora.
Tokenization is the process of splitting text into words or subwords.
Stopword removal eliminates common words like 'the' and 'is' from text.
Lemmatization reduces words to their base or root form.
TF-IDF stands for Term Frequency Inverse Document Frequency.
Cosine similarity measures the angle between two vectors in vector space.
A chatbot is a program designed to simulate conversation with humans.
Deep learning uses neural networks with many layers to learn from data.
Part of speech tagging labels words as nouns, verbs, adjectives, and so on.
Information retrieval is the process of finding relevant documents from a collection.
"""


class TFIDFChatbot:
    def __init__(self, corpus):
        self.lemmatizer = WordNetLemmatizer()
        self.sentences = nltk.sent_tokenize(corpus.strip())
        self.vectorizer = TfidfVectorizer(stop_words='english')
        self.tfidf_matrix = self.vectorizer.fit_transform(self.sentences)

        self.greetings = {
            r"hi|hello|hey":           "Hello! Ask me anything about NLP.",
            r"bye|exit|quit":          "Goodbye! Good luck with your assignment!",
            r"who are you|your name":  "I'm TFIDFBot — an NLP retrieval chatbot.",
            r"thank":                  "You're welcome!",
            r"how are you":            "I'm doing great! Ready to talk NLP.",
            r"what can you do":        "I can answer questions about NLP topics like tokenization, BERT, sentiment analysis, and more!",
        }

    def preprocess(self, text):
        tokens = nltk.word_tokenize(text.lower())
        return " ".join(self.lemmatizer.lemmatize(t) for t in tokens)

    def respond(self, user_input):
        for pattern, reply in self.greetings.items():
            if re.search(pattern, user_input.lower()):
                return reply

        query_vec = self.vectorizer.transform([user_input])
        similarities = cosine_similarity(query_vec, self.tfidf_matrix).flatten()
        best_idx = np.argmax(similarities)
        best_score = similarities[best_idx]

        if best_score < 0.1:
            return "I don't have enough info on that. Try asking about NLP topics."
        return self.sentences[best_idx]


bot = TFIDFChatbot(corpus)
print("TFIDFBot ready! Ask NLP questions. Type 'bye' to exit.\n")

while True:
    user_input = input("You: ")
    if not user_input.strip():
        continue
    response = bot.respond(user_input)
    print(f"Bot: {response}\n")
    if re.search(r"bye|exit|quit", user_input.lower()):
        break
```

```python

!pip install transformers torch -q

from transformers import pipeline
import re


print("Loading model... (first run may take a minute)\n")
qa_pipeline = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")


context = """
Natural Language Processing (NLP) is a subfield of artificial intelligence that focuses
on the interaction between computers and human language. NLP enables computers to read,
understand, and generate human language in a meaningful way. Common NLP tasks include
text classification, sentiment analysis, named entity recognition, machine translation,
and question answering.

Machine learning plays a major role in modern NLP. Deep learning models such as
transformers have revolutionized the field. BERT, which stands for Bidirectional Encoder
Representations from Transformers, was introduced by Google in 2018. BERT is pre-trained
on large text corpora and can be fine-tuned for many NLP tasks.

Tokenization is the process of splitting text into smaller units called tokens.
Lemmatization reduces words to their base form. For example, the word 'running' becomes
'run' after lemmatization. Stopwords are common words like 'the', 'is', and 'a' that are
often removed during text preprocessing.

TF-IDF stands for Term Frequency Inverse Document Frequency. It is a numerical statistic
used to reflect the importance of a word in a document relative to a collection of documents.
Cosine similarity is a metric used to measure how similar two documents are by calculating
the cosine of the angle between their vector representations.

A chatbot is a software application designed to simulate conversation with humans.
Retrieval-based chatbots use predefined responses, while generative chatbots generate
responses using deep learning models. Sentiment analysis is the use of NLP to identify
and extract subjective information from text, such as opinions and emotions.
"""


class QuestionAnsweringBot:
    def __init__(self, pipeline, context):
        self.pipeline = pipeline
        self.context = context

        self.greetings = {
            r"hi|hello|hey":          "Hello! Ask me any question about NLP.",
            r"bye|exit|quit":         "Goodbye! Good luck with your assignment!",
            r"who are you|your name": "I'm QABot — a DistilBERT-powered question answering bot.",
            r"thank":                 "You're welcome!",
            r"how are you":           "I'm doing great! Ready to answer your NLP questions.",
            r"what can you do":       "I can answer questions based on an NLP knowledge base using DistilBERT!",
        }

    def respond(self, question):
        for pattern, reply in self.greetings.items():
            if re.search(pattern, question.lower()):
                return reply

        result = self.pipeline(question=question, context=self.context)
        answer = result['answer']
        score  = result['score']

        if score < 0.1:
            return "I'm not confident about that. Try rephrasing or ask something about NLP."

        confidence = round(score * 100, 1)
        return f"{answer}  (confidence: {confidence}%)"


bot = QuestionAnsweringBot(qa_pipeline, context)
print("QABot ready! Ask questions about NLP. Type 'bye' to exit.\n")

while True:
    question = input("You: ")
    if not question.strip():
        continue
    answer = bot.respond(question)
    print(f"Bot: {answer}\n")
    if re.search(r"bye|exit|quit", question.lower()):
        break
```

```python

!pip install -q sumy


from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer
from sumy.summarizers.lsa import LsaSummarizer


text = """
Natural Language Processing (NLP) is a field of Artificial Intelligence that focuses on the interaction between computers and humans using natural language.
It includes tasks such as translation, summarization, and sentiment analysis.
NLP helps machines understand and generate human language effectively.
It is widely used in chatbots, voice assistants, and recommendation systems.
"""


parser = PlaintextParser.from_string(text, Tokenizer("english"))


summarizer = LsaSummarizer()


summary_sentences = summarizer(parser.document, 2)


print("Original Text:\n", text)
print("\nSummary:\n")

for sentence in summary_sentences:
    print(sentence)
```

```python

import torch
from transformers import MarianMTModel, MarianTokenizer

print(f"PyTorch version: {torch.__version__}")


def load_model(src_lang, tgt_lang):
    model_name = f"Helsinki-NLP/opus-mt-{src_lang}-{tgt_lang}"
    print(f"Loading model: {model_name}...")
    tokenizer = MarianTokenizer.from_pretrained(model_name)
    model     = MarianMTModel.from_pretrained(model_name)
    print("Model loaded!\n")
    return model, tokenizer


def translate(text, model, tokenizer):
    tokens = tokenizer([text], return_tensors="pt", padding=True, truncation=True)
    output = model.generate(**tokens)
    return tokenizer.decode(output[0], skip_special_tokens=True)


model, tokenizer = load_model("en", "fr")


sentences = [
    "Natural Language Processing is a subfield of artificial intelligence.",
    "Machine translation converts text from one language to another.",
    "Deep learning models have revolutionized the field of NLP.",
    "BERT is a transformer model pre-trained on large text corpora.",
    "Sentiment analysis determines whether text is positive or negative.",
]


print("=" * 60)
print("       MACHINE TRANSLATION  (English → French)")
print("=" * 60 + "\n")

for i, sentence in enumerate(sentences, 1):
    translation = translate(sentence, model, tokenizer)
    print(f"[{i}] Original   : {sentence}")
    print(f"    Translated : {translation}")
    print()

print("=" * 60)


print("\nEnter your own sentence to translate (or 'quit' to exit):\n")
while True:
    user_input = input("You (English): ").strip()
    if not user_input:
        continue
    if user_input.lower() == 'quit':
        print("Goodbye!")
        break
    result = translate(user_input, model, tokenizer)
    print(f"Bot (French) : {result}\n")
```
